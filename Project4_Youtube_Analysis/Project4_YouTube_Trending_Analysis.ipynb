```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("USvideos.csv")
df.head()
```

```python
df.columns
df.shape
```

```python
top = df.sort_values("views", ascending=False).head(10)
top[["title","views"]]
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x="views", y="title", data=top)
plt.show()
```
